# Python vs. Other Programming Languages

**1. Interpreted vs Compiled**

- Python → Interpreted (runs line by line)
- C/C++/Java → Compiled (converted to machine code first)
- Python is easier to debug but generally slower than compiled languages.

**2. Dynamically Typed vs Statically Typed**

- Python → No need to declare data types
- Java/C++ → Must declare type explicitly

```python
x = 10
x = "Hello"   # Allowed
```

In statically typed languages, this would cause an error.

**3. Indentation-Based Syntax**

- Python uses indentation instead of {}.
- Indentation is part of the syntax (not optional formatting).

**4. Simpler & More Readable**

- Python code looks closer to plain English.
- Designed for clarity and developer productivity.

**5. Rich Data Ecosystem**

- Python dominates in:

    - Data Analytics
    - Data Science
    - Machine Learning
    - Automation
- Large ecosystem: pandas, numpy, matplotlib, sklearn, etc.

# Python vs. SQL

**1. Declarative vs Procedural**

- SQL → Declarative
    - You describe what you want.
    - Database decides how to compute it.

- Python → Procedural
    - You define how to compute it step-by-step.
    
**2. SQL Works on Tables. Python Works on Objects.**

- SQL : Tables, Rows, Columns, Database
- Python - Lists, Dictionaries, Dataframes, Objects

Even Dataframe is technically an object.

**3. Set-Based vs Flexible Logic**

- SQL operates on entire datasets automatically.
- Python allows:
    - Loops
    - Custom conditions
    - Complex transformations
    - Automation
    - APIs

Python gives more control.

**4. Mental Model Shift**

- SQL → “Give me the final result.”
- Python → “Let me build this step-by-step.”

**5. Purpose Difference**

- SQL is best for retrieving and aggregating structured data.
- Python is best for processing, transforming, automating, and analyzing data.

## Assignment Evaluation Order & Reference Behavior in Python

1. **Variables Store References**

   - Variables do not store raw values.
   - They store references to objects in memory.
   - Multiple variables can reference the same object.

---
2. **Multiple Assignment Works via Tuple Packing/Unpacking**

   - No temporary variable is created during multiple assignment in Python.
   - RHS is completely evaluated first and stored in a tuple. No assignment happens until RHS is evaluated completely.
   - Then values are assigned from left to right.
   - Earlier assignments can mutate objects.
   - Later assignments can override variable references.

---
3. **Mutable Objects Can Cause Side Effects**

   - Lists, dicts, and sets are mutable.
   - Modifying one reference affects all references.
   - Reassignment changes the reference, not the object.

### Accessing Elements

- Lists use indexing to access individual elements.

```python
nums = [10, 20, 30, 40]

nums[0]   # 10
nums[2]   # 30
```

- Negative Indexing to start indexing from last.

- Start index is included, End index is excluded.

```python
nums = [10, 20, 30, 40, 50]

nums[1:4]   # [20, 30, 40] From 2nd element till 4th element
nums[:3]    # [10, 20, 30] From start till 3rd element
nums[2:]    # [30, 40, 50] From 3rd element till end
```

### Difference Operation

- The difference operation returns elements that are present in one set but not in another.
- Can be done using the minus operator or using the difference method. 

| Operation            | Meaning                           |
| -------------------- | --------------------------------- |
| `A - B`              | Elements in A but not in B        |
| `A.difference(B)`    | Same as `A - B`                   |
| `A.difference(B, C)` | Remove elements of B and C from A |
| Commutative?         | ❌ No                              |


```python
print(A - B)   # difference
print(B - A)   # difference
```

```python
A = {1, 2, 3, 4}
B = {3, 4, 5,6}
C = {2,7}

result1 = A.difference(B)
print(result1)

result2 = A.difference(B, C)
print(result2)
```

# Conditional Statements

- Conditional statements control the flow of execution of a program based on logic.

- They allow the program to make decisions based on conditions.

## 1. if Statement

- The if statement executes code only if the condition is True.
- If the condition is False, the code block does not run.

```python
age = 20

if age >= 18:
    print("Adult")
```

## 2. if-else Statement

- Used when there are two possible outcomes.

```python
age = 16

if age >= 18:
    print("Adult")
else:
    print("Minor")
```

## 3. if-elif-else Statement

- Used when there are multiple conditions.
- Python evaluates conditions top to bottom.

```python
score = 85

if score >= 90:
    print("Grade A")
elif score >= 75:
    print("Grade B")
elif score >= 60:
    print("Grade C")
else:
    print("Grade D")
```

## 4. Nested Conditions

- A condition inside another condition.

```python
age = 25
citizen = True

if age >= 18:
    if citizen:
        print("Eligible to vote")
```

## 5. Logical Operators

- Logical operators allow combining multiple conditions.

| Operator | Meaning                             |
| -------- | ----------------------------------- |
| and      | Both conditions must be True        |
| or       | At least one condition must be True |
| not      | Reverses the condition              |


```python
age = 25
has_id = True

if age >= 18 and has_id:
    print("Entry allowed")
```

## 6. Comparison Operators

- Used to compare values.

| Operator | Meaning                  |
| -------- | ------------------------ |
| ==       | Equal to                 |
| !=       | Not equal to             |
| >        | Greater than             |
| <        | Less than                |
| >=       | Greater than or equal to |
| <=       | Less than or equal to    |


```python
x = 10

if x == 10:
    print("Value is 10")
```

## 7. Ternary Conditional (Short Form if-else)
- A compact way to write if-else.

```python
age = 20

status = "Adult" if age >= 18 else "Minor"
```

# Loops

- Loops allow a program to repeat a block of code multiple times.
- They are commonly used to iterate over collections of data such as lists, tuples, sets, dictionaries, and strings.
- Python provides two main types of loops:
    - for loop
    - while loop

## 1. for Loop

- A for loop is used to iterate over a sequence (list, tuple, string, dictionary, etc.).

#### Example: Loop through a list

```python
nums = [1, 2, 3, 4]

for num in nums:
    print(num)
```

#### Example: Loop through a string

```python
text = "Python"

for char in text:
    print(char)
```

## 2. range() Function

- The range() function generates a sequence of numbers.

```python
for i in range(5):
    print(i)
```

- Range with start and step

```python
for i in range(1, 10, 2):
    print(i)
```

## 3. while Loop

- A while loop runs as long as a condition is True.
- The condition must eventually become False, otherwise the loop will run indefinitely.

```python
count = 0

while count < 5:
    print(count)
    count += 1
```

## 4. break Statement

- The break statement stops the loop immediately.

```python
for num in range(10):
    if num == 5:
        break
    print(num)
```

## 5. continue Statement

- The continue statement skips the current iteration and moves to the next one.

```python
for num in range(5):
    if num == 2:
        continue
    print(num)
```

## 6. enumerate()

- enumerate() allows access to both index and value while looping.

```python
fruits = ["apple", "banana", "mango"]

for index, fruit in enumerate(fruits):
    print(index, fruit)
```

## 7. zip()

- zip() allows iterating over multiple sequences simultaneously.

```python
names = ["Priya", "Rahul"]
scores = [90, 85]

for name, score in zip(names, scores):
    print(name, score)
```

## 8. Looping Through Dictionaries

- Looping through dictionaries can be done using items() method which gives access to both key and value at once.


```python
person = {"name": "Priya", "age": 22}

for key, value in person.items():
    print(key, value)
```

| while       | Run loop while condition is True |
| break       | Exit loop early                  |
| continue    | Skip current iteration           |

- Loops allow repeated execution of code.

# List Comprehensions

- A list comprehension provides a compact way to create lists by applying an expression to each element in an iterable.
- It is often used as a shorter and more readable alternative to loops.

## 1. Basic Syntax

```python
[expression for item in iterable]
```

```python
nums = [1, 2, 3, 4]

squares = [x**2 for x in nums]

print(squares)
```

## 2. Equivalent for Loop
- List comprehension replaces this common pattern:

```python
nums = [1, 2, 3, 4]

squares = []

for x in nums:
    squares.append(x**2)
```

- Using list comprehension:

```python
squares = [x**2 for x in nums]
```

This is **shorter** and **more readable**.

## 3. List Comprehension with Condition

### Syntax:

```python
[expression for item in iterable if condition]
```

**Example: Get even numbers.**

```python
nums = [1, 2, 3, 4, 5, 6]

evens = [x for x in nums if x % 2 == 0]

print(evens)
```

## 4. Conditional Expression (if–else)

- List comprehensions can include if–else expressions.

```python
nums = [1, 2, 3, 4]

labels = ["Even" if x % 2 == 0 else "Odd" for x in nums]

print(labels)
```

## 5. Working with Strings

**Example: Convert characters to uppercase.**

```python
text = "python"

uppercase_chars = [char.upper() for char in text]

print(uppercase_chars)
```

## 6. Nested List Comprehension

- Used when iterating over nested structures.

**Example: Flatten a list.**

```python
matrix = [[1, 2], [3, 4], [5, 6]]

flattened = [num for row in matrix for num in row]

print(flattened)
```

## Summary Table

| Pattern        | Example                                       |
| -------------- | --------------------------------------------- |
| Basic          | `[x for x in nums]`                           |
| Transformation | `[x**2 for x in nums]`                        |
| With condition | `[x for x in nums if x % 2 == 0]`             |
| With if–else   | `["Even" if x%2==0 else "Odd" for x in nums]` |
| Nested         | `[num for row in matrix for num in row]`      |

- List comprehensions provide a compact way to generate lists.
- They combine iteration, transformation, and filtering in one line.
- Often replace simple loops used for building lists.
- Commonly used in data cleaning and preprocessing tasks.

# Functions

Functions help in:
- Organizing code
- Avoiding repetition
- Improving readability

## 1. Defining a Function

- Functions are defined using the **def** keyword.

```python
def greet():
    print("Hello!")
```

- Calling the function

```python
greet()
```

## 2. Functions with Parameters

- Functions can accept inputs (parameters).

```python
def greet(name):
    print("Hello", name)

greet("Priya")
```

- Functions can return results using the return keyword.

```python
def add(a, b):
    return a + b

result = add(3, 5)
print(result)
```

- Functions can have default values for parameters.

```python
def greet(name="Guest"):
    print("Hello", name)

greet()
greet("Priya")
```

## 5. Lambda Functions

- A lambda function is a small anonymous function written in one line.
- Can take any number of arguments
- Contain only one expression
- Return value automatically
- Often used for short temporary functions inside map(), sorted(), reduce(), filter()

**Syntax**

```python
lambda arguments: expression
```

```python
square = lambda x: x**2

square(4)
```

- Even though square references it, the function itself has the internal name *lambda*, which is why it is called anonymous.

## 6. Variable-Length Arguments

####  *args (multiple positional arguments)

- Allows a function to accept multiple inputs.

```python
def add_numbers(*args):
    return sum(args)

add_numbers(1, 2, 3, 4)
```

#### **kwargs (multiple keyword arguments)

- Accepts named arguments.

```python
def print_info(**kwargs):
    for key, value in kwargs.items():
        print(key, value)

print_info(name="Priya", age=22)
```

# Exception Handling

- Exception handling allows a program to handle runtime errors gracefully without stopping execution abruptly.
- Python provides the try, except, else, and finally blocks for handling exceptions.

## 1. Basic try-except

- Code that may produce an error is placed inside a try block.
- If an error occurs, the except block is executed.

```python
try:
    x = 10 / 0
except ZeroDivisionError:
    print("Cannot divide by zero")
```

## 2. Handling Multiple Exceptions

- A program may encounter different types of errors.

```python
try:
    num = int("abc")
except ValueError:
    print("Invalid number")
except TypeError:
    print("Type error occurred")
```

## 3. Using else

- The else block runs only if no exception occurs.

```python
try:
    x = 10 / 2
except ZeroDivisionError:
    print("Error occurred")
else:
    print("Division successful")
```

## 4. Using finally

- The finally block always executes, regardless of whether an exception occurs.
- Often used for: closing files, releasing resources

```python
try:
    x = 10 / 2
except ZeroDivisionError:
    print("Error")
finally:
    print("Execution completed")
```

## 5. Catching All Exceptions

- Sometimes a generic exception is used.

```python
try:
    x = int("hello")
except Exception as e:
    print("Error:", e)
```

- Here *e* stores the error message.

## Common Python Exceptions

| Exception         | Cause                           |
| ----------------- | ------------------------------- |
| ZeroDivisionError | Division by zero                |
| ValueError        | Invalid value type              |
| TypeError         | Invalid operation between types |
| IndexError        | List index out of range         |
| KeyError          | Dictionary key not found        |
