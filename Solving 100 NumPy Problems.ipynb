## 15. Create a 2D array with 1 on borders and 0 on inside.

```python
nos = np.zeros((5,5))
nos
```

```python
nos[0,:] = 1 # top row
nos[-1,:] = 1 # bottom row
nos[:,0] = 1 # first column
nos[:,-1] = 1 # last column
nos
```

```python
# More generic solution 1
size = 9
ones = np.ones((size,size))
print(ones)
zeros = np.zeros((size-2,size -2))
print(zeros)
```

```python
ones[1:size-1,1:size-1] = zeros
ones
```

```python
# More generic solution 2
size = 5
zeros = np.zeros((size,size))
zeros
```

```python
zeros[[0,-1]] = 1 # top and bottom row
zeros[:,[0,-1]] = 1 # first and last column
zeros
```

## 18. Create a 5 x 5 matrix with values 1,2,3,4 just below the diagonal.

## 19. Create a 8 x 8 matrix and fill it with checkerboard pattern.

```python
checkerboard = np.zeros((8,8))
checkerboard
```

```python
arr = np.ones((4,4))
arr
```

```python
checkerboard[::2,::2] = arr #(every second row, every second column starting from index 0)
checkerboard
```

```python
checkerboard[1::2,1::2] = arr #(every second row, second column starting from index 1)
checkerboard
```

```python
# Another way (thinking in terms of odd index and even index)
checkerboard[::2, 1::2] = 1
checkerboard[1::2, ::2] = 1
checkerboard
```

## 21. Create a checkerboard (8 x 8) matrix using the tile function.

```python
a = np.array([0,1])

odd_matrix = np.tile(a,4)
even_matrix = np.tile(a[::-1], 4)

checkerboard = np.tile(np.hstack((odd_matrix, even_matrix)),4).reshape(8,8)
checkerboard
```

```python
# Another way using just tile function

np.tile(np.array([[0,1],[1,0]]), (4,4))
```

## 33. How to get the dates of yesterday, today and tomorrow?

```python
# This is using the datetime library
from datetime import datetime, timedelta

today = datetime.now()
print(today)

yesterday = today - timedelta(days = 1)
print(yesterday)

tomorrow = today + timedelta(days = 1)
print(tomorrow)
```

```python
# Using NumPy
today = np.datetime64('today')
print(today)

yesterday = today - np.timedelta64(1, 'D')
print(yesterday)

tomorrow = today + np.timedelta64(1, 'D')
print(tomorrow)
```
