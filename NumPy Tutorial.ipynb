### Reorganizing Arrays

```python
# Reshaping Arrays
```

Reshaping means changing the dimensions (shape) of an array without changing its data.

You can rearrange how data is viewed (like turning a row into a column, or a 1D list into a matrix).

```python
arr.reshape(new_shape)
```

The total number of elements must stay the same.

```python
a = np.array([1,2,3,4,5,6])
a.shape
```

```python
a.reshape(1,6)
```

```python
a.reshape(6,1)
```

NumPy can figure out one dimension automatically if you put -1 there.

```python
a.reshape(2,-1)
```

```python
a.reshape(-1,3)
```

```python
a.reshape(3,1,2)
```

#### Boolean Masking and Advanced Indexing

Boolean masking means using True/False values to filter or select elements from an array.
You create a Boolean array (mask) of the same shape as your data,
and NumPy returns only the elements where the mask is True.

```python
a = np.array([1,2,3,4])

# Boolean mask
a > 1
```

```python
# Pass the Boolean mask as a filter
a[a > 1]
```

```python
filedata [filedata > 100]
```

```python
filedata[filedata > 50]
```

So Boolean masking works across any shape — it flattens the result into 1D when filtered.

```python
# Return a bollean array when any of the elements meet the condition
np.any(filedata > 50, axis = 1) # along the rows
```

```python
np.any(filedata > 50, axis = 0) # along the columns
```

```python
# Combining Multiple Conditions using Logical Operators
(filedata > 50) & (filedata < 100)
```

```python
(~((filedata > 50) & (filedata < 100)))
```

```python
filedata[(~((filedata > 50) & (filedata < 100)))]
```

```python
~(filedata[((filedata > 50) & (filedata < 100))])
```

The ~ operator in NumPy performs a bitwise NOT, meaning it inverts the bits of each integer.

For integers, the operation is equivalent to:

 
x = - (x + 1)
